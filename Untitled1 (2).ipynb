```python
T = [ 20 , 25 , 30 , 35 , 40 ]
C = [ 40 , 55 , 70 , 85 , 100 ]
print(T)
print(C) 
```

```python
import matplotlib.pyplot as plt
```

```python
plt.plot(T, C , marker='o')

```

```python
plt.scatter(T, C) 
plt.xlabel("Température(C°)")
plt.ylabel("Croissance bactérienne")
```

```python
import numpy as np
T = np.array([20, 30, 35, 40])
C = np.array([40, 70, 85, 100])
```

```python
from sklearn.linear_model import LinearRegression
X = T.reshape(-1, 1)
Y = C

model = LinearRegression()
model.fit(X, Y)

a = model.coef_[0]
b = model.intercept_

print("a =", a)
print("b =", b)
```

```python
print(f"l'équation de régression est : Y = {a}X + ({b})")
```

```python
Y_pred = model.predict(X)
print("Valeurs prédites:", Y_pred)
```

```python
plt.scatter(T, C, color="blue", label="Données réelles")
plt.plot(T, Y_pred, color="red",label="Droite de régression")

plt.xlabel("Température (°C)")
plt.ylabel("Croissance bactérienne")
plt.title("Régression linéaire")

plt.legend()
plt.show()
```

```python
temperature_test = np.array([[50]])

prediction = model.predict(temperature_test)

print("prédiction a 50°C :",prediction[0])
```
